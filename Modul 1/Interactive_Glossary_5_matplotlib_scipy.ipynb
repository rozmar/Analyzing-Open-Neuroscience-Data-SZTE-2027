# Creating DataFrame for dictionaries

We can create a `DataFrame` from a list of `dict`. In this case, values with the same key belong to the same column. If a key from any of the dictionaries is missing, there will be an empty value (nan).


```python
# We can create a DataFrame from a list of dictionaries
demo = pd.DataFrame([
    {'cell_id': 'c1', 'SW': 32.76, 'HW': 0.165, 'cell_type': 'FSI'},
    {'cell_id': 'c2', 'SW': 49.83, 'HW': 0.194, 'cell_type': 'FSI'},
    {'cell_id': 'c3', 'SW': 24.23, 'HW': 0.112, 'cell_type': 'FSI'},
    {'cell_id': 'c4', 'SW': 0.91, 'HW': 0.41, 'cell_type': 'Pyr'},
    {'cell_id': 'c5', 'SW': 0.307, 'HW': 0.37, 'cell_type': 'Pyr'},
    {'cell_id': 'c6', 'SW': 0.36, 'HW': 0.51, 'cell_type': 'Pyr'},
    {'cell_id': 'c7', 'SW': 1.57, 'HW': 0.424, 'cell_type': 'Pyr'},
    {'cell_id': 'c8', 'SW': 13.23, 'HW': 0.207, 'cell_type': 'RSI'},
    {'cell_id': 'c9', 'SW': 28.69, 'HW': 0.277, 'cell_type': 'RSI'},
    {'cell_id': 'c10', 'SW': 8.47, 'HW': 0.243, 'cell_type': 'RSI'},
])
# Let's set the cell_id column as index
demo = demo.set_index("cell_id")
demo.head(10)
```

**Aggregations** in pandas are functions that reduce many values to one summary value per a column, row or group.\
Here are a few examples:

- count() – number of (non-missing) values.

- sum() – total of values.

- mean() – arithmetic average.

- median() – middle value (50th percentile).

- min() / max() – smallest / largest value.

- std() / var() – standard deviation / variance (spread).

- nunique() – number of unique values.

- first() / last() – first / last non-missing entry.

- describe() – quick summary (count, mean, std, min, quartiles, max).

```python
# Column-wise aggregations
print("mean SW:", demo["SW"].mean())
demo[["SW","HW"]].std()
```

#Missing values (NaN)

pandas represents missing data as **NaN**. There are a few methods that help deal with them.
- `isna()` creates a bolean mask (True where the value is NA\
- `fillna()` replaces the missing (NA) value (in acopy not the original DataFrame)\
- `dropna()` removes missing rows and columns with missing data


```python
# We can create a DataFrame from a list of dictionaries
demo_with_missing = pd.DataFrame([
    {'cell_id': 'c1', 'SW': 32.76, 'HW': 0.165, 'cell_type': 'FSI'},
    {'cell_id': 'c2', 'SW': 49.83, 'HW': 0.194, 'cell_type': 'FSI'},
    {'cell_id': 'c3', 'SW': 24.23, 'HW': 0.112, 'cell_type': 'FSI'},
    {'cell_id': 'c4', 'HW': 0.41, 'cell_type': 'Pyr'},
    {'cell_id': 'c5', 'SW': 0.307, 'HW': 0.37, 'cell_type': 'Pyr'},
    {'cell_id': 'c6', 'SW': 0.36, 'HW': 0.51, 'cell_type': 'Pyr'},
    {'cell_id': 'c7', 'SW': 1.57, 'cell_type': 'Pyr'},
    {'cell_id': 'c8', 'SW': 13.23, 'HW': 0.207, 'cell_type': 'RSI'},
    {'cell_id': 'c9', 'SW': 28.69, 'HW': 0.277, 'cell_type': 'RSI'},
    {'cell_id': 'c10', 'HW': 0.243, 'cell_type': 'RSI'},
    {'cell_id': 'c11', 'SW': 0.36, 'HW': 0.51},
])
# Let's set the cell_id column as index
demo_with_missing = demo_with_missing.set_index("cell_id")
demo_with_missing.head(10)
```

```python
demo_filled = demo_with_missing.copy()   #create a copy of incomplete_df which we are going to fill
demo_filled
```

```python
demo_filled['SW'] = demo_filled['SW'].fillna(demo_filled['SW'].mean())
demo_filled['HW'] = demo_filled['HW'].fillna(demo_filled['HW'].mean())
demo_filled = demo_filled.dropna()
demo_filled
```

<h2> Matplotlib Basics </h2>

We will be using pyplot.subplots (<code>plt.subplots()</code>) to create our beautiful data visualizations. Calling
<code>plt.subplots()</code> gives back two objecs that are conventionally called a "figure" and an "axes". To represent these we will create two variables, **fig** and **ax**.
    
**fig:** You can think of the figure as the frame of your plot.
* You can resize, reshape the frame but you cannot draw on it.
* On a single notebook or a script, you can have multiple figures.
* Each figure can have multiple subplots.
       
**ax:** an axes is your metaphorical canvas to draw on. Or rephrasing, it is the blank sheet you can plot and hold your data.
* An axes object can only belong to one figure.

## scipy.signal
### Spectral analysis (FFT)

```python
from scipy import signal
```

Calculate the power spectrum of the two signals in the `DataFrame` using the `welch` method.

```python
f1, Pxx_1 = signal.welch(oscill['v1'].to_numpy(), 1000, nperseg=256, scaling='spectrum')
f2, Pxx_2 = signal.welch(oscill['v2'].to_numpy(), 1000, nperseg=256, scaling='spectrum')
fig, ax = plt.subplots(1, 2, figsize=(20, 5))
ax[0].plot(f1, Pxx_1)
ax[0].set_xlim(0, 120)
ax[1].plot(f2, Pxx_2)
ax[1].set_xlim(0, 40)
plt.show()
```

### Filtering
We filter the first signal using a second order, Butterworth bandpass filter between `[10,20] Hz`. When we plot the filtered signal, we see the almost pure 15 Hz oscillation. It can be seen on the power spectrum of the filtered signal.

```python
b, a = signal.butter(2, [10, 20], 'bandpass', fs=1000)
filt_v1 = signal.filtfilt(b, a, oscill['v1'].to_numpy())
filt_f, filt_Pxx = signal.welch(filt_v1, 1000, nperseg=256, scaling='spectrum')

fig, ax = plt.subplots(1, 2, figsize=(20, 5))
ax[0].plot(oscill['t'], filt_v1)
ax[1].plot(filt_f, filt_Pxx)
ax[1].set_xlim(0, 100)
plt.show()
```

Filter the second signal using a second order, Butterworth lowpass filter below `30 Hz`.

```python
b, a = signal.butter(2, 20, 'low', fs=1000)
filt_v2 = signal.filtfilt(b, a, oscill['v2'].to_numpy())
filt_f2, filt_Pxx2 = signal.welch(filt_v2, 1000, nperseg=256, scaling='spectrum')

fig, ax = plt.subplots(1, 2, figsize=(20, 5))
ax[0].plot(oscill['t'], filt_v2)
ax[1].plot(filt_f2, filt_Pxx2)
ax[1].set_xlim(0, 100)
plt.show()
```

### Pearson correlation
We can calculate the Pearson's correlation coefficient with the corresponding p-value of two set of points.

```python
x = np.random.uniform(0, 10, size=(50,))
y1 = 1.5 * x + np.random.normal(0, 1, size=(50,))
y2 = np.random.uniform(0, 10, size=(50,))
y3 = -2 * x + np.random.normal(0, 1, size=(50,))

r1, p1 = stats.pearsonr(x, y1)
r2, p2 = stats.pearsonr(x, y2)
r3, p3 = stats.pearsonr(x, y3)

fig, ax = plt.subplots(1, 3, figsize=(20, 5))
ax[0].scatter(x, y1)
ax[0].set_title("r =" + str(np.round(r1, 3)) + " p= " + str(np.round(p1, 3)))
ax[1].scatter(x, y2)
ax[1].set_title("r =" + str(np.round(r2, 3)) + " p= " + str(np.round(p2, 3)))
ax[2].scatter(x, y3)
ax[2].set_title("r =" + str(np.round(r3, 3)) + " p= " + str(np.round(p3, 3)))
plt.show()
```

### t-test
We can compare the means of two random samples. If both of them follows normal distribution, we can use t-test.

```python
x1 = np.random.normal(5, 1, size=(50,))
x2 = np.random.normal(5.1, 1, size=(35,))
x3 = np.random.normal(10, 1, size=(25,))

s1 = stats.ttest_ind(x1, x2)
s2 = stats.ttest_ind(x1, x3)

print(s1)
print(s2)
```
