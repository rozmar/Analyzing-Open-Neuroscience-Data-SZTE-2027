<h2>1. The notebook and the runtime are not the same thing</h2>

<p>A Colab notebook is a <code>.ipynb</code> file. It contains the <b>code cells</b>, <b>text cells</b>, and often the outputs that you can see on the page.</p>

<p>The <b>runtime</b> is the temporary computer that actually executes your Python code.</p>

<p>A useful biology analogy:</p>

<ul>
<li><b>Notebook = laboratory protocol / lab notebook</b>: it contains the instructions.</li>
<li><b>Runtime = the experiment currently running on the bench</b>: it contains the current state of the experiment.</li>
</ul>

<p>Saving the protocol does not preserve the exact current state of the experiment.  
Similarly, saving a Colab notebook does <b>not</b> permanently save all Python variables that currently exist in memory.</p>
</div>


```python
# This code creates three variables in the CURRENT runtime.

species = "mouse"
cell_type = "PV"
spike_count = 17

print(species)
print(cell_type)
print(spike_count)
```


<div style="border-left: 3px solid #000; padding: 1px; padding-left: 10px; background: #F0FAFF;">
<p>After the cell above has been executed, Python remembers these names in the current runtime:</p>

<pre>
species
cell_type
spike_count
</pre>

<p>The code itself is written in the notebook, but the <b>current values</b> of the variables are stored in the runtime's memory.</p>
</div>


```python
# We can use the variables in a later cell because they already exist in memory.

print(cell_type + " cells")
print("Spike count:", spike_count)
```

<h2>2. Cells are executed in the order you run them</h2>

<p>The notebook is displayed from top to bottom, but Python does not automatically execute it from top to bottom.</p>

<p>Python only knows which cells you have <b>actually run</b> in the current runtime.</p>

<p>This means that a variable can appear to be defined in a cell above, but it still does not exist unless that cell has been executed.</p>
</div>


```python
message = "This variable now exists in the runtime."
print(message)
```

```python
# Skip this cell
message = "This variable does not exist yet."
```

```python
# This works only if the previous cell has already been run.
print(message)
```


<div style="background: #FFF7D6; border-radius: 3px; padding: 10px;">
<p><b>Try this:</b></p>

<ol>
<li>Run both cells above.</li>
<li>Change the value of <code>message</code> in the first cell.</li>
<li>Run only the second cell.</li>
</ol>

<p>What happens?</p>

<p>The second cell still uses the <b>old value</b>, because editing a cell does not execute it.  
You must run the edited cell before the runtime knows about the new value.</p>
</div>



<div style="border-left: 3px solid #000; padding: 1px; padding-left: 10px; background: #F0FAFF;">
<h3>Running the same cell again can change the runtime state</h3>

<p>A cell does not have to be executed only once. You can run it repeatedly.</p>
</div>


```python
# First initialize the counter.
counter = 0
print("counter =", counter)
```

```python
# Run THIS cell several times.
counter = counter + 1
print("counter =", counter)
```


<div style="border-left: 3px solid #000; padding: 1px; padding-left: 10px; background: #F0FAFF;">
<p>Every time the second cell is run, it uses the value currently stored in the runtime and adds 1.</p>

<p>This is why the visible order of cells and the <b>execution history</b> are not always the same thing.</p>

<p>If a notebook starts behaving strangely, one useful strategy is to restart the runtime and then run the notebook again from the top.</p>
</div>



<div style="border-left: 3px solid #000; padding: 1px; padding-left: 10px; background: #F0FAFF;">
<h2>3. What happens when the runtime is restarted?</h2>

<p>If the runtime is restarted, disconnected, or deleted, the current Python memory is cleared.</p>

<p>This means that variables such as:</p>

<pre>
species
cell_type
spike_count
counter
</pre>

<p>will disappear from memory.</p>

<p>The notebook itself can still contain the cells that create them, so you can recreate the variables simply by running the relevant cells again.</p>
</div>

<h2>4. Files in the Colab runtime</h2>

<p>The runtime also has its own file system.</p>

<p>In Colab, the usual working directory is:</p>

<pre>/content</pre>

<p>You can think of this as the temporary working folder of the cloud computer.</p>
</div>


```python
# os is part of Python's standard library and helps us work with files and folders.

import os

print("Current working directory:")
print(os.getcwd())

print("\nFiles and folders currently here:")
print(os.listdir())
```


<div style="border-left: 3px solid #000; padding: 1px; padding-left: 10px; background: #F0FAFF;">

<pre>

!ls
</pre>

<p>These are not normal Python functions. The exclamation mark tells Jupyter/Colab to run a shell command on the underlying computer.</p>
</div>

<h3>Saving a file inside the runtime</h3>

<p>Let's create a small result file. This file will be saved in the runtime's current working directory.</p>
</div>


```python
# A very small CSV-like result file

results = [
    "cell,membrane_potential_mV",   # This code block create a Python list called "results".
    "cell_1,-68.2",                 # Each item in the list is a string will be one line in a CSV file.
    "cell_1,-68.2",
    "cell_2,-71.5",
    "cell_3,-65.9"
]

# Open a file called "example_results.csv".
# "w" means write mode:
# - if the file does not exist, Python creates it (This is what will happen now.) if it already exists, Python overwrites its contents

with open("example_results.csv", "w") as file:
    file.write("\n".join(results))

print("Saved to:")
print(os.path.abspath("example_results.csv"))
```

```python
# Let's look at where is our example_results file saved

!pwd
!ls
```

```python
print("Current working directory:")
print(os.getcwd())

print("\nFiles and folders currently here:")
print(os.listdir())
```

```python
# Read the file back to check what we saved.

with open("example_results.csv", "r") as file:
    contents = file.read()

print(contents)
```


<div style="background: #FFF7D6; border-radius: 3px; padding: 10px;">
<p><b>Important:</b></p>

<p>A file saved in <code>/content</code> is usually <b>temporary</b>.  
It may disappear when the runtime is deleted or reset.</p>

<p>This is different from a file saved in Google Drive or downloaded to your own computer.</p>
</div>

<h2>7. Saving the notebook itself</h2>

<p>The notebook (<code>.ipynb</code>) and the files created by your Python code are two different things.</p>

<p>To keep your own editable copy of a notebook, you can use:</p>

<pre>File → Save a copy in Drive</pre>

<p>This is especially important when you opened a notebook from GitHub or from a course link.</p>

<p>Your saved notebook contains the code and text cells, and it may also contain displayed outputs.  
However, it does <b>not</b> permanently preserve the current runtime memory.</p>
</div>



<div style="border-left: 3px solid #000; padding: 1px; padding-left: 10px; background: #F0FAFF;">
<h2>8. Connecting Google Drive to the runtime</h2>

<p>If you want Python to read or write files directly in your Google Drive, you can <b>mount</b> Drive.</p>

<p>Running the next cell will ask you to authorize access.</p>
</div>


```python
from google.colab import drive

drive.mount("/content/drive")
```


<div style="border-left: 3px solid #000; padding: 1px; padding-left: 10px; background: #F0FAFF;">
<p>After mounting, your Google Drive is usually available under:</p>

<pre>/content/drive/MyDrive/</pre>

<p>You can now access files in Drive using normal Python file paths.</p>
</div>


```python
# This cell is useful AFTER Drive has been mounted.

drive_root = "/content/drive/MyDrive"

print("Drive folder exists:", os.path.exists(drive_root))

if os.path.exists(drive_root):
    print("Some items in MyDrive:")
    print(os.listdir(drive_root)[:10])
```


<div style="border-left: 3px solid #000; padding: 1px; padding-left: 10px; background: #F0FAFF;">
<h2>9. Saving a result permanently to Google Drive</h2>

<p>Let's copy our example result file from the temporary runtime into a folder in Google Drive.</p>

<p>The code below creates a folder called <code>Python_course_demo</code> in <code>MyDrive</code> if it does not already exist.</p>
</div>


```python
import shutil

drive_folder = "/content/drive/MyDrive/Python_course_demo"
os.makedirs(drive_folder, exist_ok=True)

source_file = "example_results.csv"
destination_file = os.path.join(drive_folder, "example_results.csv")

shutil.copy(source_file, destination_file)

print("Saved to Drive:")
print(destination_file)
```


<div style="border-left: 3px solid #000; padding: 1px; padding-left: 10px; background: #F0FAFF;">
<p>The difference is important:</p>

<ul>
<li><code>/content/example_results.csv</code> → temporary runtime file</li>
<li><code>/content/drive/MyDrive/Python_course_demo/example_results.csv</code> → file stored in Google Drive</li>
</ul>

<p>If the runtime disappears, the copy stored in Google Drive should remain there.</p>
</div>

<h2>10. Opening notebooks from GitHub</h2>

<p>Course notebooks can also be stored on GitHub.</p>

<p>In Colab you can open them using:</p>

<pre>File → Open notebook → GitHub</pre>

<p>Opening a notebook from GitHub gives you access to that notebook, but it does <b>not</b> mean that your changes are automatically saved back to the original GitHub repository.</p>

<p>For normal course work, the safest workflow is:</p>

<ol>
<li>Open the course notebook from GitHub.</li>
